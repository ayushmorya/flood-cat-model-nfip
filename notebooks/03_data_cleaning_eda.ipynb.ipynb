```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.max_columns", 100)
```

```python
df = pd.read_csv("nfip_claims_harris_tx.csv")
print(f"Rows: {len(df):,}   Columns: {df.shape[1]}")
df.head()
```

```python
df.info()
```

```python
#missing values check
key_columns = [
    "yearOfLoss",
    "buildingDamageAmount",
    "netBuildingPaymentAmount",
    "buildingPropertyValue",
    "buildingReplacementCost",
    "totalBuildingInsuranceCoverage",
    "floodZoneCurrent",
    "causeOfDamage",
]
missing = df[key_columns].isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(1)
pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
```

```python
#duplicates

print(f"Duplicate IDs: {df['id'].duplicated().sum()}")
print(f"Fully duplicate rows: {df.duplicated().sum()}")
```

```python
#value ranges
df[["buildingDamageAmount", "netBuildingPaymentAmount", "buildingPropertyValue", "buildingReplacementCost", "totalBuildingInsuranceCoverage"]].describe()
```

```python
#apply cleaning (with reasoning in the comments):
before = len(df)
df_clean = df.copy()

#building value must be positive to compute damage ratio
df_clean = df_clean[df_clean["buildingPropertyValue"] > 0]

#Damage and payment amounts can't be negative (Data entry errors)
df_clean = df_clean[df_clean["buildingDamageAmount"].fillna(0) >= 0]
df_clean = df_clean[df_clean["netBuildingPaymentAmount"].fillna(0) >= 0]

#Fill missing flood zone with an explicit label rather than dropping rows
df_clean["floodZoneCurrent"] = df_clean["floodZoneCurrent"].fillna("UNKNOWN")

after = len(df_clean)
print(f"Rows before cleaning: {before:,}")
print(f"Rows after cleaning: {after:,}")
print(f"Rows removed: {before - after:,} ({(before - after) / before:.1%})")
```

```python
df_clean.info()
```

```python
#claims per year chart:

claims_per_year = df_clean.groupby("yearOfLoss").size()
ax = claims_per_year.plot(kind="bar", color="#2b6cb0")
ax.set_title("Harris County, TX - NFIP Claims by Year")
ax.set_xlabel("Year of Loss")
ax.set_ylabel("Number of Claims")
plt.tight_layout()
plt.show()
```

```python
#claims by flood zone

zone_counts = df_clean["floodZoneCurrent"].value_counts().head(15)
ax = zone_counts.plot(kind = "bar", color = "#2c7a7b")
ax.set_title("Claims by Flood Zone (Top 15)")
plt.tight_layout()
plt.show()
```

```python
#damage distributions

positive_damage = df_clean[df_clean["buildingDamageAmount"] > 0]["buildingDamageAmount"]
ax = positive_damage.plot(kind="hist", bins=60, color="#c05621", log=True)
ax.set_title("Distribution of Building Damage Amounts (log scale)")
plt.tight_layout()
plt.show()
positive_damage.describe()
```

```python
#damage ratio by flood zone

df_clean["damage_ratio"] = (df_clean["buildingDamageAmount"] / df_clean["buildingPropertyValue"]).clip(upper=2.0)
ratio_by_zone = (df_clean[df_clean["buildingDamageAmount"] > 0]
    .groupby("floodZoneCurrent")["damage_ratio"]
    .agg(["mean", "median", "count"])
    .sort_values("count", ascending=False).head(10))
ratio_by_zone
```

```python
df_clean.to_csv("nfip_claims_harris_tx_clean.csv", index=False)
print(f"Saved {len(df_clean):,} cleaned rows")
```
